# ClaimsPilot ? Sprint 2 NLP Error Analysis

Development-set error analysis using MLflow predictions.

**MLflow run:** `ac72240a02ab4f6cb793d7ff3614b42d`

**Dataset:** `synthetic_claims_v3.csv` (220 synthetic development claims).

**Important:** These results are not independent held-out evaluation.

```
import pandas as pd
from IPython.display import display

errors = [{'claim_id': 'C156', 'true_label': 'inconsistent', 'predicted_label': 'uncertain', 'predicted_parts': [], 'score': 0.0}, {'claim_id': 'C213', 'true_label': 'inconsistent', 'predicted_label': 'uncertain', 'predicted_parts': ['front_bumper', 'door'], 'score': 0.5}, {'claim_id': 'C214', 'true_label': 'inconsistent', 'predicted_label': 'uncertain', 'predicted_parts': ['rear_bumper', 'headlight'], 'score': 0.5}, {'claim_id': 'C215', 'true_label': 'inconsistent', 'predicted_label': 'uncertain', 'predicted_parts': ['front_bumper'], 'score': 0.5}]
df = pd.DataFrame(errors)
display(df)
print('Total errors:', len(df))
```

## Error categories

**C156 ? Missing synonym:** The English phrase `front light` refers to the headlight but is not extracted.

**C213, C214, C215 ? Partial overlap:** The expected labels are `inconsistent`, while the current Jaccard-threshold policy predicts `uncertain`.

These disagreements require confirmation of the team's annotation and consistency policy.

## Development results

- Macro F1: **0.8982**
- Inconsistent recall: **0.9612**

- Misclassified claims: **4 / 220**
